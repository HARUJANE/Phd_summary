```python
import os
import subprocess

# === 路径设置 ===
input_dir = r"D:\github\PhD\nagahma\data\feng\ICA_txt"
exe_path = r"D:\github\PhD\nagahma\data\code\curvature\batchVModeler.exe"
param_file = r"D:\github\PhD\nagahma\data\code\curvature\param_5d_L3.txt"
output_root = r"D:\github\PhD\nagahma\data\code\curvature\output"

# === 固定参数 ===
mode_number = "1030"
flag1 = "1"
flag2 = "1"
flag3 = "1"
scale = "1.0"

# === 遍历每个 txt 文件 ===
for fn in os.listdir(input_dir):
    if not fn.endswith(".txt"):
        continue

    input_file = os.path.join(input_dir, fn)
    base_name = os.path.splitext(fn)[0]  # 去掉后缀
    out_dir = os.path.join(output_root, base_name)
    os.makedirs(out_dir, exist_ok=True)

    # 拼接命令
    cmd = [
        exe_path,
        mode_number,
        input_file,
        param_file,
        flag1, flag2, flag3,
        out_dir + "\\",  # 确保结尾有反斜杠
        scale
    ]

    print("正在运行：", " ".join(cmd))
    # 执行命令
    subprocess.run(cmd)

print("全部文件已处理完成。")
```

```python
import os
import shutil

output_root = r"D:\github\PhD\nagahma\data\code\curvature\output"
sigma0_suffix = "_sigma0"
target_contains = "0_vals.txt"  # 宽松匹配

save_dir = r"D:\github\PhD\nagahma\data\code\curvature\output_curvature"
os.makedirs(save_dir, exist_ok=True)

count = 0
for model_folder in os.listdir(output_root):
    model_path = os.path.join(output_root, model_folder)
    if not os.path.isdir(model_path):
        continue

    sigma0_folders = [f for f in os.listdir(model_path)
                      if sigma0_suffix in f and os.path.isdir(os.path.join(model_path, f))]

    for sigma0_folder in sigma0_folders:
        sigma0_path = os.path.join(model_path, sigma0_folder)

        for file in os.listdir(sigma0_path):
            if target_contains in file:
                src_file = os.path.join(sigma0_path, file)

                # 去掉文件名中从 "_spline" 开始的部分，只保留 "_spline" 之前+ .txt
                if "_spline" in file:
                    base_name = file.split("_spline")[0] + ".txt"
                else:
                    # 如果没有 "_spline"，就用原来的名字
                    base_name = os.path.splitext(file)[0] + ".txt"

                dst_file = os.path.join(save_dir, base_name)

                shutil.copy2(src_file, dst_file)
                count += 1
                #print(f"[OK] {src_file} → {dst_file}")

print(f"总共复制了 {count} 个文件到 {save_dir}")
```
